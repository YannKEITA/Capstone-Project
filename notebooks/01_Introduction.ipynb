# 🏡 Projet Kaggle – Prédiction des prix de l'immobilier

## 🎯 Objectif du projet

Ce projet s'inscrit dans le cadre du célèbre challenge Kaggle **"House Prices - Advanced Regression Techniques"**.  
L'objectif est de prédire le **prix de vente de maisons** situées à Ames (Iowa, USA) à partir d’un ensemble riche de **caractéristiques techniques** et **environnementales** (79 variables au total).

Nous cherchons à construire un modèle de régression robuste capable de généraliser au mieux sur les données test.

---

## 📂 Description des données

Le dataset est composé de deux fichiers CSV principaux :

- `train.csv` – 1460 observations, 79 variables explicatives + `SalePrice` (la cible)
- `test.csv` – 1459 observations, même structure mais sans la variable cible

Les variables incluent des informations sur :
- la **superficie** (habitable, sous-sol, garage, etc.)
- l’**année de construction ou rénovation**
- la **qualité** de matériaux et de finition
- des informations **catégorielles** (quartier, type de maison, finition cuisine...)
- le **nombre de salles de bain**, chambres, garages, etc.

---

## 🔧 Tâches réalisées dans ce projet

1. **Exploration des données** (EDA)
2. **Nettoyage** et **prétraitement**
3. **Sélection de variables pertinentes**
4. **Optimisation d’hyperparamètres** (Optuna)
5. **Empilement (stacking)** de plusieurs modèles
6. **Analyse de l’importance des variables (SHAP)**
7. **Évaluation des performances**
8. **Analyse des résidus**

---

## 🧠 Objectif final

- Maximiser les performances du modèle selon la métrique **Root Mean Squared Log Error (RMSLE)** utilisée par Kaggle.
- Créer un modèle **généralisable** et **interprétable** pour des prédictions fiables sur de nouvelles données.

---

📌 Ce projet est un cas concret de **régression supervisée** appliquée à un jeu de données réel, idéal pour mettre en pratique les compétences en **data science**, **machine learning**, et **analyse de données**.
