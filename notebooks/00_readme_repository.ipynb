# 🏠 House Prices Prediction - Kaggle Regression Challenge

Ce projet a été réalisé dans le cadre d’un exercice pratique de machine learning supervisé en régression, en utilisant les données du concours **"House Prices - Advanced Regression Techniques"** sur [Kaggle](https://www.kaggle.com/competitions/house-prices-advanced-regression-techniques).

---

## 📌 Objectif

Prédire le prix de vente (`SalePrice`) de maisons résidentielles à Ames, Iowa, à partir de caractéristiques structurelles, temporelles et géographiques.

---

## 🧪 Méthodologie

Le projet a été structuré en 8 étapes :

1. **Input**  
   - Données fiables issues de Kaggle.
   - Données nettoyées, structurées, et bien annotées.

2. **Data Analysis (DA)**  
   - Exploration approfondie avec de nombreux plots.
   - Identification des variables clés et des outliers.

3. **Preprocessing**  
   - Encodage des variables catégorielles, imputation des valeurs manquantes.
   - Transformation logarithmique de la cible (`SalePrice`).

4. **Feature Selection**  
   - Sélection manuelle guidée par la corrélation, le bon sens métier, et des méthodes automatiques (RFE, RF importance).
   - Sélection finale de 30 variables.

5. **Modelling**  
   - Test de plusieurs modèles (Ridge, Lasso, XGBoost).
   - Optimisation des hyperparamètres avec **Optuna**.
   - Combinaison via **StackingRegressor**.

6. **Feature Analysis**  
   - Analyse de l’importance des variables avec **SHAP**.
   - Visualisation de la convergence des trials Optuna.

7. **Evaluation**  
   - Validation croisée (RMSE) pour chaque modèle.
   - Comparaison des performances individuelles vs stacking.

8. **Résidual Analysis**  
   - Distribution des résidus.
   - Visualisation des erreurs vs prédictions.

---

## 🏁 Résultats

- **RMSE stacking final : `0.1405`**
- **Modèle final :** Stacking(Ridge + Lasso + XGBoost)
- Visualisation des features les plus influentes avec SHAP.
