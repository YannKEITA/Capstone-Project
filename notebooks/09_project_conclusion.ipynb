## ✅ Conclusion du Projet de Prédiction – House Prices

Ce projet visait à prédire le prix de vente de maisons à Ames, Iowa, en utilisant un ensemble de données publiques fournies par Kaggle.

---

### 📌 Objectif atteint :
Nous avons construit un modèle performant avec une RMSE finale de **0.1405** grâce à :
- Une sélection rigoureuse des variables,
- L’optimisation d’hyperparamètres avec **Optuna**,
- L’agrégation de modèles via **Stacking** (Ridge, Lasso, XGBoost),
- Une évaluation croisée robuste et une analyse des résidus complète.

---

### 📚 Ce que nous avons appris :
- Le **nettoyage des données** et le **choix des bonnes features** sont déterminants.
- Le **stacking** permet de combiner les forces de différents modèles pour maximiser la performance.
- **SHAP** permet une explication fine et interprétable des prédictions.

---

### 🚀 Perspectives :
- Intégrer davantage de variables catégorielles ou transformer les features existantes (ex. : interactions).
- Tester des modèles plus complexes comme **LightGBM**, **CatBoost**, ou des réseaux de neurones.
- Soumettre le modèle final sur Kaggle pour évaluer sa compétitivité.
